# Functions for Reusable Data Tasks
Task 4 - Day 4

```python
# 1. Mean
def calc_mean(nums):
    return sum(nums) / len(nums)

marks = [45, 67, 89, 34, 92, 58]
print(calc_mean(marks))
```

```python
# 2. Maximum without built-in max()
def calc_max(nums):
    result = nums[0]
    for n in nums:
        if n > result:
            result = n
    return result

print(calc_max(marks))
```

```python
# 3. Minimum without built-in min()
def calc_min(nums):
    result = nums[0]
    for n in nums:
        if n < result:
            result = n
    return result

print(calc_min(marks))
```

```python
# 4. Percentage - function with two parameters
def calc_percentage(part, whole):
    return (part / whole) * 100

print(calc_percentage(45, 60))
```

```python
# 5. Celsius to Fahrenheit - single parameter, returns a value
def celsius_to_fahrenheit(c):
    return (c * 9 / 5) + 32

print(celsius_to_fahrenheit(30))
```

```python
# 6. Grade from marks - parameter with a default value
def grade_from_marks(marks, passing=40):
    if marks < passing:
        return "Fail"
    elif marks >= 90:
        return "A+"
    elif marks >= 75:
        return "A"
    elif marks >= 60:
        return "B"
    else:
        return "C"

print(grade_from_marks(82))
print(grade_from_marks(30))
print(grade_from_marks(55, passing=50))
```

```python
# 7. Discount price - default parameter, overridden on call
def discount_price(price, pct=10):
    return price - (price * pct / 100)

print(discount_price(500))
print(discount_price(500, 20))
```

## Using the functions together

```python
# Applying a reusable function across a list
for m in marks:
    print(m, "->", grade_from_marks(m))
```

```python
# Combining multiple functions in one calculation
prices = [500, 1200, 750, 300]
for p in prices:
    print(p, "->", discount_price(p, 15))
```

## Interview questions

**What is a function in Python?**
A reusable block of code defined with `def` that performs a specific task. It can take input through parameters and send output back using `return`, so the same logic doesn't need to be rewritten every time it's needed.

**What is the difference between a parameter and an argument?**
A parameter is the placeholder name listed in the function definition (e.g. `nums` in `def calc_mean(nums):`). An argument is the actual value passed in when the function is called (e.g. `calc_mean(marks)` — `marks` is the argument).

**Why are functions useful in data science projects?**
They stop the same calculation from being copy-pasted everywhere, which means a fix or improvement only needs to happen in one place. They also make code easier to test and read, and let the same logic be applied consistently across many different datasets.
